## 1. Pengenalan Exploratory Data Analysis (EDA)

Exploratory Data Analysis (EDA) adalah proses untuk memahami dan mengeksplorasi data sebelum melakukan analisis lebih lanjut.

EDA digunakan untuk melihat karakteristik data, menemukan pola, mengetahui persebaran data, dan mendapatkan gambaran awal dari sebuah dataset.

### Tujuan EDA

Tujuan dari EDA adalah:
1. Memahami struktur dataset.
2. Mengetahui karakteristik setiap variabel.
3. Mengetahui nilai statistik dasar.
4. Menemukan pola atau informasi awal dari data.
5. Membantu menentukan analisis yang dapat dilakukan selanjutnya.

### Workflow EDA

Workflow EDA secara sederhana adalah:

Dataset → Memahami Data → Statistik Deskriptif → Visualisasi → Menemukan Pola → Insight

### Jenis Analisis EDA

**1. Univariate**

Analisis yang dilakukan pada satu variabel. Contohnya melihat distribusi panjang sepal menggunakan histogram.

**2. Bivariate**

Analisis yang dilakukan pada dua variabel untuk melihat hubungan antarvariabel.

**3. Multivariate**

Analisis yang menggunakan lebih dari dua variabel untuk melihat hubungan atau pola yang lebih kompleks.

```python
iris = load_iris(as_frame=True)
df = iris.frame

df = df.rename(columns={
    'sepal length (cm)': 'sepal_length',
    'sepal width (cm)': 'sepal_width',
    'petal length (cm)': 'petal_length',
    'petal width (cm)': 'petal_width'
})

df['species'] = df['target'].map({
    0: 'setosa',
    1: 'versicolor',
    2: 'virginica'
})

df.head()
```

```python
kolom_numerik = [
    'sepal_length',
    'sepal_width',
    'petal_length',
    'petal_width'
]
```

```python
df[kolom_numerik].mean().round(2)
```

```python
df[kolom_numerik].median().round(2)
```

```python
df[kolom_numerik].mode().iloc[0]
```

```python
df[kolom_numerik].min()
```

```python
df[kolom_numerik].max()
```

```python
df[kolom_numerik].std().round(2)
```

```python
df[kolom_numerik].quantile([0.25, 0.50, 0.75]).round(2)
```

```python
statistical_summary = pd.DataFrame({
    'Mean': df[kolom_numerik].mean(),
    'Median': df[kolom_numerik].median(),
    'Modus': df[kolom_numerik].mode().iloc[0],
    'Minimum': df[kolom_numerik].min(),
    'Maksimum': df[kolom_numerik].max(),
    'Standar Deviasi': df[kolom_numerik].std(),
    'Q1': df[kolom_numerik].quantile(0.25),
    'Q2': df[kolom_numerik].quantile(0.50),
    'Q3': df[kolom_numerik].quantile(0.75)
}).round(2)

statistical_summary
```

```python
with pd.ExcelWriter('Statistical Summary.xlsx') as writer:
    statistical_summary.to_excel(
        writer,
        sheet_name='Statistical Summary'
    )

    df['species'].value_counts().to_excel(
        writer,
        sheet_name='Species Count'
    )

    df.head(10).to_excel(
        writer,
        sheet_name='Data Sample',
        index=False
    )

print("Statistical Summary.xlsx berhasil dibuat!")
```
