Analisis Dataset Iris untuk menemukan root cause, alternatif solusi, dan rekomendasi berbasis data.

## 1. Finding the Root Cause

Proses identifikasi atau penyortiran bunga Iris berpotensi kurang efektif karena beberapa spesies memiliki karakteristik fisik yang berdekatan.

### Root Cause
Pemilihan fitur yang kurang tepat dapat membuat proses identifikasi menjadi kurang efektif. Hasil analisis menunjukkan bahwa fitur petal memberikan perbedaan antarspesies yang lebih jelas, sedangkan Versicolor dan Virginica memiliki beberapa karakteristik yang relatif berdekatan.

## 2. Data Evidence

- Dataset memiliki 150 observasi.
- Terdapat 3 spesies yang dianalisis.
- Setiap spesies memiliki 50 observasi.
- Petal Length memiliki feature separation terbesar yaitu 4.09.
- Korelasi Petal Length dan Petal Width mencapai 0.96.
- Rata-rata Petal Length Virginica adalah 5.55 cm, sedangkan Versicolor adalah 4.26 cm.

## 3. Alternative Solutions

- Memprioritaskan fitur Petal
- Menggunakan kombinasi Sepal + Petal
- Mengembangkan sistem klasifikasi otomatis
- Menambahkan pemeriksaan manual

## 4. Data-Driven Recommendation

- Prioritaskan Petal Length dan Petal Width
- Gunakan kombinasi beberapa fitur
- Kembangkan sistem klasifikasi otomatis

## 5. Stakeholder Consultation

**Q:** Mengapa proses identifikasi kita bermasalah?

**A:** Karena beberapa spesies memiliki karakteristik fisik yang berdekatan sehingga tidak semua fitur dapat membedakannya dengan jelas.

**Q:** Apa penyebab utamanya?

**A:** Penyebab utamanya adalah belum adanya prioritas fitur berdasarkan kemampuan masing-masing fitur dalam membedakan spesies.

**Q:** Spesies mana yang paling bermasalah?

**A:** Versicolor dan Virginica perlu mendapat perhatian karena beberapa karakteristiknya relatif berdekatan.

**Q:** Fitur apa yang paling membantu?

**A:** Petal Length menjadi fitur paling menonjol berdasarkan feature separation sebesar 4.09.

**Q:** Apa bukti datanya?

**A:** Dataset memiliki 150 observasi, 3 spesies, dan korelasi Petal Length dengan Petal Width sebesar 0.96.

**Q:** Apa yang harus dilakukan?

**A:** Memprioritaskan fitur petal dan menggunakan kombinasi beberapa fitur untuk kasus yang lebih sulit.

**Q:** Apa solusi jangka panjangnya?

## 6. Final Recommendation

Prioritaskan Petal Length dan Petal Width sebagai fitur utama, gunakan kombinasi beberapa fitur untuk kasus yang sulit, lalu kembangkan sistem klasifikasi otomatis sebagai tahap lanjutan.

## 7. Expected Impact

## 8. Conclusion

Masalah utama bukan hanya perbedaan antarspesies, tetapi bagaimana memilih karakteristik yang paling informatif untuk membantu proses identifikasi.

```python
from sklearn.datasets import load_iris
import pandas as pd

iris = load_iris(as_frame=True)

df = iris.frame.copy()

df = df.rename(columns={
    "sepal length (cm)": "Sepal Length",
    "sepal width (cm)": "Sepal Width",
    "petal length (cm)": "Petal Length",
    "petal width (cm)": "Petal Width",
    "target": "Target"
})

df["Species"] = df["Target"].map(
    dict(enumerate(iris.target_names))
)

df.head()
```

```python
mean_by_species = df.groupby("Species")[
    ["Sepal Length",
     "Sepal Width",
     "Petal Length",
     "Petal Width"]
].mean()

mean_by_species
```

```python
feature_separation = (
    mean_by_species.max()
    - mean_by_species.min()
).sort_values(ascending=False)

feature_separation
```

```python
correlation = df[
    ["Sepal Length",
     "Sepal Width",
     "Petal Length",
     "Petal Width"]
].corr()

correlation
```
