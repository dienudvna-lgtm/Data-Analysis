```python
# ============================================================
# WEEK 7 - EXPLORATORY DATA ANALYSIS LANJUTAN
# Dataset: Iris
# ============================================================

# Jika library belum tersedia, jalankan baris ini sekali:
# !pip install pandas numpy matplotlib scikit-learn openpyxl reportlab

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from reportlab.lib import colors
from reportlab.lib.pagesizes import A4
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib.enums import TA_CENTER
from reportlab.platypus import (
    SimpleDocTemplate, Paragraph, Spacer, Table, TableStyle,
    PageBreak, Image
)
from reportlab.lib.units import cm

print("=" * 70)
print("WEEK 7 - EXPLORATORY DATA ANALYSIS LANJUTAN")
print("=" * 70)

# ============================================================
# 1. LOAD DATASET IRIS
# ============================================================

iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=[
        "sepal_length",
        "sepal_width",
        "petal_length",
        "petal_width"
    ]
)

df["species"] = pd.Categorical.from_codes(
    iris.target,
    iris.target_names
)

print("\n[1] DATASET BERHASIL DIMUAT")
print("Jumlah baris :", len(df))
print("Jumlah kolom :", len(df.columns))
print("\n5 data pertama:")
display(df.head())

# ============================================================
# 2. INFORMASI DATASET
# ============================================================

print("\n" + "=" * 70)
print("INFORMASI DATASET")
print("=" * 70)

print("\nUkuran dataset:", df.shape)

print("\nMissing value:")
print(df.isnull().sum())

print("\nJumlah data duplikat:")
print(df.duplicated().sum())

print("\nJumlah masing-masing species:")
print(df["species"].value_counts().sort_index())

# ============================================================
# 3. STATISTIK DESKRIPTIF
# ============================================================

numeric_cols = [
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width"
]

stats = df[numeric_cols].describe().T

stats["median"] = df[numeric_cols].median()

stats = stats[
    [
        "count",
        "mean",
        "median",
        "std",
        "min",
        "25%",
        "50%",
        "75%",
        "max"
    ]
]

print("\n" + "=" * 70)
print("STATISTIK DESKRIPTIF")
print("=" * 70)

display(stats.round(2))

# ============================================================
# 4. TREND ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("HARI SENIN - TREND ANALYSIS")
print("=" * 70)

# Iris tidak memiliki tanggal.
# Oleh karena itu tanggal dibuat sebagai simulasi untuk latihan.

trend_df = df.copy()

trend_df["Date"] = pd.date_range(
    start="2026-01-01",
    periods=len(trend_df),
    freq="D"
)

monthly = (
    trend_df
    .set_index("Date")
    .resample("MS")["petal_length"]
    .mean()
    .reset_index()
)

monthly.rename(
    columns={"petal_length": "average_petal_length"},
    inplace=True
)

monthly["growth_percentage"] = (
    monthly["average_petal_length"]
    .pct_change() * 100
)

display(monthly.round(2))

# ---------------- LINE CHART ----------------

plt.figure(figsize=(10, 5))

plt.plot(
    monthly["Date"],
    monthly["average_petal_length"],
    marker="o"
)

plt.title(
    "Trend Analysis - Rata-rata Petal Length per Bulan"
)

plt.xlabel("Bulan")
plt.ylabel("Rata-rata Petal Length")
plt.grid(True, alpha=0.3)

plt.tight_layout()

line_path = "Trend_Line_Chart.png"
plt.savefig(line_path, dpi=150)
plt.show()

# ---------------- AREA CHART ----------------

plt.figure(figsize=(10, 5))

plt.fill_between(
    monthly["Date"],
    monthly["average_petal_length"],
    alpha=0.4
)

plt.plot(
    monthly["Date"],
    monthly["average_petal_length"]
)

plt.title(
    "Trend Analysis - Area Chart Petal Length"
)

plt.xlabel("Bulan")
plt.ylabel("Rata-rata Petal Length")

plt.grid(True, alpha=0.3)

plt.tight_layout()

area_path = "Trend_Area_Chart.png"
plt.savefig(area_path, dpi=150)
plt.show()

print("""
INSIGHT TREND:

1. Data Iris tidak mempunyai atribut waktu asli.
2. Tanggal pada analisis ini merupakan tanggal simulasi.
3. Grafik digunakan untuk menunjukkan cara melakukan analisis trend.
4. Karena tanggalnya simulasi, hasil ini tidak dapat dianggap sebagai
   trend waktu nyata atau seasonality bisnis.
""")

# ============================================================
# 5. SEGMENTASI DATA
# ============================================================

print("\n" + "=" * 70)
print("HARI SELASA - SEGMENTASI DATA")
print("=" * 70)

segment_df = df.copy()

# Segmentasi 1: Species
segment_1 = (
    segment_df
    .groupby("species", observed=True)
    .size()
    .reset_index(name="jumlah")
)

print("\nSEGMENTASI 1 - SPECIES")
display(segment_1)

# Segmentasi 2: Sepal Length
segment_df["sepal_length_segment"] = pd.cut(
    segment_df["sepal_length"],
    bins=3,
    labels=["Rendah", "Sedang", "Tinggi"],
    include_lowest=True
)

segment_2 = (
    segment_df
    .groupby("sepal_length_segment", observed=True)
    .size()
    .reset_index(name="jumlah")
)

print("\nSEGMENTASI 2 - SEPAL LENGTH")
display(segment_2)

# Segmentasi 3: Sepal Width
segment_df["sepal_width_segment"] = pd.cut(
    segment_df["sepal_width"],
    bins=3,
    labels=["Rendah", "Sedang", "Tinggi"],
    include_lowest=True
)

segment_3 = (
    segment_df
    .groupby("sepal_width_segment", observed=True)
    .size()
    .reset_index(name="jumlah")
)

print("\nSEGMENTASI 3 - SEPAL WIDTH")
display(segment_3)

# Segmentasi 4: Petal Length
segment_df["petal_length_segment"] = pd.cut(
    segment_df["petal_length"],
    bins=3,
    labels=["Rendah", "Sedang", "Tinggi"],
    include_lowest=True
)

segment_4 = (
    segment_df
    .groupby("petal_length_segment", observed=True)
    .size()
    .reset_index(name="jumlah")
)

print("\nSEGMENTASI 4 - PETAL LENGTH")
display(segment_4)

# Segmentasi 5: Petal Width
segment_df["petal_width_segment"] = pd.cut(
    segment_df["petal_width"],
    bins=3,
    labels=["Rendah", "Sedang", "Tinggi"],
    include_lowest=True
)

segment_5 = (
    segment_df
    .groupby("petal_width_segment", observed=True)
    .size()
    .reset_index(name="jumlah")
)

print("\nSEGMENTASI 5 - PETAL WIDTH")
display(segment_5)

# ---------------- STACKED BAR ----------------

segment_plot = pd.crosstab(
    segment_df["species"],
    segment_df["petal_length_segment"]
)

segment_plot.plot(
    kind="bar",
    stacked=True,
    figsize=(10, 5)
)

plt.title("Stacked Bar - Segmentasi Petal Length berdasarkan Species")
plt.xlabel("Species")
plt.ylabel("Jumlah Data")
plt.xticks(rotation=0)

plt.tight_layout()

stacked_path = "Segment_Stacked_Bar.png"
plt.savefig(stacked_path, dpi=150)
plt.show()

# ---------------- CLUSTERED BAR ----------------

cluster_plot = pd.crosstab(
    segment_df["species"],
    segment_df["petal_width_segment"]
)

cluster_plot.plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Clustered Bar - Segmentasi Petal Width berdasarkan Species")
plt.xlabel("Species")
plt.ylabel("Jumlah Data")
plt.xticks(rotation=0)

plt.tight_layout()

clustered_path = "Segment_Clustered_Bar.png"
plt.savefig(clustered_path, dpi=150)
plt.show()

# Simpan ringkasan segmentasi
segment_summary = pd.DataFrame({
    "Segmentasi": [
        "Species",
        "Sepal Length",
        "Sepal Width",
        "Petal Length",
        "Petal Width"
    ],
    "Keterangan": [
        "Pengelompokan berdasarkan jenis species",
        "Pengelompokan berdasarkan ukuran sepal length",
        "Pengelompokan berdasarkan ukuran sepal width",
        "Pengelompokan berdasarkan ukuran petal length",
        "Pengelompokan berdasarkan ukuran petal width"
    ]
})

segment_summary.to_csv(
    "Segment_Analysis_Summary.csv",
    index=False
)

print("""
INSIGHT SEGMENTASI:

1. Dataset terbagi menjadi tiga species.
2. Masing-masing species memiliki jumlah data yang sama.
3. Sepal length dapat dikelompokkan menjadi rendah, sedang, dan tinggi.
4. Sepal width dapat dikelompokkan menjadi rendah, sedang, dan tinggi.
5. Petal length dan petal width memberikan segmentasi yang lebih
   informatif untuk membedakan karakteristik species.
""")

# ============================================================
# 6. OUTLIER ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("HARI RABU - OUTLIER & PATTERN ANALYSIS")
print("=" * 70)

outlier_results = []

for col in numeric_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_mask = (
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    )

    outlier_count = outlier_mask.sum()

    outlier_results.append({
        "Variable": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Jumlah Outlier": outlier_count
    })

outlier_summary = pd.DataFrame(outlier_results)

print("\nHASIL DETEKSI OUTLIER:")
display(outlier_summary.round(2))

outlier_summary.to_csv(
    "Outlier_Analysis_Summary.csv",
    index=False
)

# ---------------- BOXPLOT ----------------

plt.figure(figsize=(10, 6))

df[numeric_cols].boxplot()

plt.title("Boxplot - Outlier Analysis")
plt.ylabel("Nilai")
plt.xticks(rotation=20)

plt.tight_layout()

boxplot_path = "Outlier_Boxplot.png"
plt.savefig(boxplot_path, dpi=150)
plt.show()

# ---------------- DISTRIBUSI ----------------

df[numeric_cols].hist(
    bins=12,
    figsize=(10, 7)
)

plt.suptitle(
    "Distribusi Variabel Numerik Dataset Iris"
)

plt.tight_layout()

distribution_path = "Outlier_Distribution.png"
plt.savefig(distribution_path, dpi=150)
plt.show()

# ---------------- SCATTER ----------------

plt.figure(figsize=(9, 6))

plt.scatter(
    df["petal_length"],
    df["petal_width"]
)

plt.title(
    "Scatter Plot - Petal Length vs Petal Width"
)

plt.xlabel("Petal Length")
plt.ylabel("Petal Width")

plt.grid(True, alpha=0.3)

plt.tight_layout()

scatter_path = "Pattern_Scatter_Plot.png"
plt.savefig(scatter_path, dpi=150)
plt.show()

correlation = df["petal_length"].corr(
    df["petal_width"]
)

print(
    "\nKorelasi Petal Length dan Petal Width:",
    round(correlation, 3)
)

print("""
INSIGHT OUTLIER & PATTERN:

1. Metode IQR digunakan untuk mendeteksi outlier.
2. Tidak semua nilai ekstrem merupakan kesalahan data.
3. Outlier harus diperiksa terlebih dahulu sebelum dihapus.
4. Scatter plot menunjukkan hubungan yang sangat kuat antara
   petal length dan petal width.
5. Dalam data bisnis nyata, outlier dapat berasal dari kesalahan
   input, transaksi khusus, atau kondisi yang tidak biasa.
""")

# ============================================================
# 7. DATA DASHBOARD
# ============================================================

print("\n" + "=" * 70)
print("HARI KAMIS - DASHBOARD")
print("=" * 70)

dashboard = df.copy()

# Tanggal simulasi
dashboard["Date"] = pd.date_range(
    start="2026-01-01",
    periods=len(dashboard),
    freq="D"
)

# Data bisnis turunan/simulasi
dashboard["Product"] = (
    dashboard["species"]
    .astype(str)
    .str.title()
)

dashboard["Category"] = (
    dashboard["species"]
    .astype(str)
    .str.title()
)

regions = ["West", "East", "North", "South"]

dashboard["Region"] = np.resize(
    regions,
    len(dashboard)
)

dashboard["Customer"] = [
    f"Customer-{i:03d}"
    for i in range(1, len(dashboard) + 1)
]

dashboard["Quantity"] = (
    dashboard["petal_width"] * 10
).round().astype(int) + 1

dashboard["Unit Price"] = (
    dashboard["petal_length"] * 10
).round().astype(int)

dashboard["Sales"] = (
    dashboard["Quantity"] *
    dashboard["Unit Price"]
)

print("\nContoh data dashboard:")
display(dashboard.head())

# ============================================================
# KPI
# ============================================================

total_sales = dashboard["Sales"].sum()
total_quantity = dashboard["Quantity"].sum()
average_sales = dashboard["Sales"].mean()

print("\nKPI DASHBOARD")
print("Total Sales    :", total_sales)
print("Total Quantity :", total_quantity)
print("Average Sales  :", round(average_sales, 2))

# ============================================================
# GRAFIK PENJUALAN
# ============================================================

monthly_sales = (
    dashboard
    .set_index("Date")
    .resample("MS")["Sales"]
    .sum()
    .reset_index()
)

plt.figure(figsize=(10, 5))

plt.plot(
    monthly_sales["Date"],
    monthly_sales["Sales"],
    marker="o"
)

plt.title("Grafik Penjualan Bulanan")
plt.xlabel("Bulan")
plt.ylabel("Sales")

plt.grid(True, alpha=0.3)

plt.tight_layout()

sales_chart_path = "Dashboard_Sales_Chart.png"
plt.savefig(sales_chart_path, dpi=150)
plt.show()

# ============================================================
# GRAFIK KATEGORI
# ============================================================

category_sales = (
    dashboard
    .groupby("Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(9, 5))

category_sales.plot(
    kind="bar"
)

plt.title("Grafik Sales berdasarkan Kategori")
plt.xlabel("Kategori")
plt.ylabel("Sales")
plt.xticks(rotation=0)

plt.tight_layout()

category_chart_path = "Dashboard_Category_Chart.png"
plt.savefig(category_chart_path, dpi=150)
plt.show()

# ============================================================
# GRAFIK WILAYAH
# ============================================================

region_sales = (
    dashboard
    .groupby("Region")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(9, 5))

region_sales.plot(
    kind="bar"
)

plt.title("Grafik Sales berdasarkan Wilayah")
plt.xlabel("Wilayah")
plt.ylabel("Sales")
plt.xticks(rotation=0)

plt.tight_layout()

region_chart_path = "Dashboard_Region_Chart.png"
plt.savefig(region_chart_path, dpi=150)
plt.show()

# ============================================================
# 8. MEMBUAT DASHBOARD.XLSX
# ============================================================

dashboard_file = "Dashboard.xlsx"

with pd.ExcelWriter(
    dashboard_file,
    engine="openpyxl"
) as writer:

    dashboard.to_excel(
        writer,
        sheet_name="Data",
        index=False
    )

    stats.round(2).to_excel(
        writer,
        sheet_name="Statistik"
    )

    monthly_sales.to_excel(
        writer,
        sheet_name="Sales Bulanan",
        index=False
    )

    category_sales.reset_index(
        name="Sales"
    ).to_excel(
        writer,
        sheet_name="Kategori",
        index=False
    )

    region_sales.reset_index(
        name="Sales"
    ).to_excel(
        writer,
        sheet_name="Wilayah",
        index=False
    )

    pd.DataFrame({
        "KPI": [
            "Total Sales",
            "Total Quantity",
            "Average Sales"
        ],
        "Value": [
            total_sales,
            total_quantity,
            average_sales
        ]
    }).to_excel(
        writer,
        sheet_name="KPI",
        index=False
    )

print("\nDashboard.xlsx berhasil dibuat.")

# ============================================================
# 9. SETUP PDF
# ============================================================

styles = getSampleStyleSheet()

title_style = ParagraphStyle(
    "CustomTitle",
    parent=styles["Title"],
    alignment=TA_CENTER,
    fontSize=20,
    spaceAfter=20
)

heading_style = ParagraphStyle(
    "CustomHeading",
    parent=styles["Heading2"],
    fontSize=14,
    spaceBefore=10,
    spaceAfter=8
)

normal_style = ParagraphStyle(
    "CustomNormal",
    parent=styles["BodyText"],
    fontSize=10,
    leading=15,
    spaceAfter=8
)

def create_pdf(filename, title, content, images=None):

    doc = SimpleDocTemplate(
        filename,
        pagesize=A4,
        rightMargin=1.5*cm,
        leftMargin=1.5*cm,
        topMargin=1.5*cm,
        bottomMargin=1.5*cm
    )

    story = []

    story.append(
        Paragraph(title, title_style)
    )

    story.append(
        Paragraph(
            "Week 7 - Exploratory Data Analysis Lanjutan",
            normal_style
        )
    )

    story.append(Spacer(1, 10))

    for item in content:

        if item.startswith("## "):

            story.append(
                Paragraph(
                    item.replace("## ", ""),
                    heading_style
                )
            )

        else:

            story.append(
                Paragraph(
                    item,
                    normal_style
                )
            )

    if images:

        for image in images:

            if os.path.exists(image):

                story.append(
                    Spacer(1, 10)
                )

                story.append(
                    Image(
                        image,
                        width=16*cm,
                        height=9*cm
                    )
                )

    doc.build(story)

# ============================================================
# 10. TREND ANALYSIS PDF
# ============================================================

trend_content = [

"## Tujuan Analisis",

"Trend Analysis digunakan untuk melihat perubahan data berdasarkan waktu.",

"## Hasil Analisis",

"Dataset Iris tidak mempunyai kolom tanggal. Oleh karena itu, tanggal dibuat secara simulasi mulai 1 Januari 2026 untuk keperluan latihan analisis trend.",

"Line Chart digunakan untuk melihat perubahan rata-rata petal length dari waktu ke waktu.",

"Area Chart digunakan untuk memberikan gambaran visual mengenai perubahan nilai dari waktu ke waktu.",

"## Insight",

"1. Data yang digunakan untuk waktu merupakan data simulasi.",

"2. Tidak dapat disimpulkan adanya trend waktu nyata dari dataset Iris.",

"3. Tidak dapat disimpulkan adanya seasonality nyata karena dataset Iris bukan time-series.",

"4. Pada dataset bisnis nyata, kolom tanggal dapat digunakan untuk melihat pertumbuhan, penurunan, dan pola musiman."

]

create_pdf(
    "Trend Analysis.pdf",
    "Trend Analysis",
    trend_content,
    [
        line_path,
        area_path
    ]
)

# ============================================================
# 11. SEGMENT ANALYSIS PDF
# ============================================================

segment_content = [

"## Tujuan Analisis",

"Segmentasi dilakukan untuk membagi data menjadi kelompok tertentu agar karakteristik data lebih mudah dianalisis.",

"## Lima Segmentasi",

"1. Species — mengelompokkan data berdasarkan jenis species.",

"2. Sepal Length — membagi ukuran sepal length menjadi rendah, sedang, dan tinggi.",

"3. Sepal Width — membagi ukuran sepal width menjadi rendah, sedang, dan tinggi.",

"4. Petal Length — membagi ukuran petal length menjadi rendah, sedang, dan tinggi.",

"5. Petal Width — membagi ukuran petal width menjadi rendah, sedang, dan tinggi.",

"## Insight",

"Segmentasi petal length dan petal width cukup informatif untuk melihat perbedaan karakteristik antar-species.",

"Pada data bisnis nyata, konsep yang sama dapat diterapkan untuk segmentasi wilayah, produk, pelanggan, atau kategori."

]

create_pdf(
    "Segment Analysis.pdf",
    "Segment Analysis",
    segment_content,
    [
        stacked_path,
        clustered_path
    ]
)

# ============================================================
# 12. OUTLIER ANALYSIS PDF
# ============================================================

outlier_content = [

"## Tujuan Analisis",

"Analisis outlier digunakan untuk menemukan nilai yang berbeda jauh dari pola umum data.",

"## Metode",

"Metode Interquartile Range (IQR) digunakan dengan batas bawah Q1 - 1.5 × IQR dan batas atas Q3 + 1.5 × IQR.",

"## Hasil",

"Boxplot digunakan untuk melihat distribusi dan kemungkinan nilai ekstrem.",

"Scatter plot digunakan untuk melihat pola hubungan antara petal length dan petal width.",

f"Korelasi antara petal length dan petal width adalah sekitar {correlation:.3f}, yang menunjukkan hubungan linear yang sangat kuat.",

"## Dampak Bisnis",

"Karena dataset Iris bukan dataset bisnis, dampak bisnis berikut bersifat hipotetis. Pada data bisnis nyata, outlier dapat memengaruhi rata-rata, laporan KPI, model analitik, dan keputusan bisnis.",

"Outlier tidak boleh langsung dihapus. Data perlu diperiksa terlebih dahulu untuk mengetahui apakah merupakan kesalahan atau kondisi yang memang valid."

]

create_pdf(
    "Outlier Analysis.pdf",
    "Outlier & Pattern Analysis",
    outlier_content,
    [
        boxplot_path,
        distribution_path,
        scatter_path
    ]
)

# ============================================================
# 13. INSIGHT REPORT PDF
# ============================================================

insight_content = [

"## Ringkasan Dataset",

"Dataset Iris terdiri dari 150 observasi dan empat variabel numerik, yaitu sepal length, sepal width, petal length, dan petal width. Dataset memiliki tiga species.",

"## 10 Insight Utama",

"1. Dataset terdiri dari 150 observasi.",

"2. Dataset memiliki tiga species: setosa, versicolor, dan virginica.",

"3. Setiap species memiliki 50 observasi sehingga distribusi kelas seimbang.",

"4. Dataset tidak memiliki missing value.",

"5. Petal length dan petal width mempunyai hubungan yang sangat kuat.",

"6. Petal length membantu membedakan karakteristik species.",

"7. Petal width juga menjadi variabel penting dalam membedakan species.",

"8. Segmentasi rendah, sedang, dan tinggi dapat digunakan untuk mengelompokkan variabel numerik.",

"9. Metode IQR dapat digunakan sebagai langkah awal untuk mendeteksi outlier.",

"10. Trend dan atribut bisnis pada dashboard merupakan simulasi karena dataset Iris tidak menyediakan data waktu maupun transaksi bisnis.",

"## Statistik Ringkas"

]

# Tabel statistik
table_data = [
    ["Variabel", "Mean", "Median", "Min", "Max"]
]

for col in numeric_cols:

    table_data.append([
        col,
        f"{df[col].mean():.2f}",
        f"{df[col].median():.2f}",
        f"{df[col].min():.2f}",
        f"{df[col].max():.2f}"
    ])

doc = SimpleDocTemplate(
    "Insight Report.pdf",
    pagesize=A4,
    rightMargin=1.5*cm,
    leftMargin=1.5*cm,
    topMargin=1.5*cm,
    bottomMargin=1.5*cm
)

story = []

story.append(
    Paragraph(
        "Insight Report",
        title_style
    )
)

story.append(
    Paragraph(
        "Week 7 - Exploratory Data Analysis Lanjutan",
        normal_style
    )
)

for item in insight_content:

    if item.startswith("## "):

        story.append(
            Paragraph(
                item.replace("## ", ""),
                heading_style
            )
        )

    else:

        story.append(
            Paragraph(
                item,
                normal_style
            )
        )

table = Table(
    table_data,
    colWidths=[
        4*cm,
        2.5*cm,
        2.5*cm,
        2.5*cm,
        2.5*cm
    ]
)

table.setStyle(
    TableStyle([
        ("BACKGROUND", (0,0), (-1,0), colors.lightgrey),
        ("GRID", (0,0), (-1,-1), 0.5, colors.grey),
        ("ALIGN", (1,1), (-1,-1), "CENTER"),
        ("FONTNAME", (0,0), (-1,0), "Helvetica-Bold"),
        ("FONTSIZE", (0,0), (-1,-1), 9),
        ("VALIGN", (0,0), (-1,-1), "MIDDLE")
    ])
)

story.append(table)

doc.build(story)

# ============================================================
# 14. BUSINESS RECOMMENDATION PDF
# ============================================================

recommendation_content = [

"## Catatan Penting",

"Dataset Iris merupakan dataset pembelajaran dan bukan dataset transaksi bisnis. Oleh karena itu, rekomendasi bisnis berikut merupakan rekomendasi metodologis yang dapat diterapkan pada dataset bisnis nyata.",

"## Rekomendasi",

"1. Prioritaskan variabel yang memiliki informasi paling kuat untuk analisis.",

"2. Gunakan segmentasi untuk memahami kelompok data dengan karakteristik yang berbeda.",

"3. Validasi outlier sebelum melakukan penghapusan atau perubahan data.",

"4. Gunakan dashboard untuk memonitor KPI dan membantu proses pengambilan keputusan.",

"5. Pada dataset bisnis nyata, sertakan atribut waktu, produk, pelanggan, wilayah, kategori, jumlah, dan nilai transaksi.",

"6. Gunakan analisis trend untuk memantau pertumbuhan dan penurunan performa.",

"7. Evaluasi hasil analisis secara berkala agar keputusan yang diambil tetap sesuai kondisi data terbaru.",

"## Prioritas Tindakan",

"Urutan tindakan yang disarankan adalah: Validasi Data → Segmentasi → Monitoring KPI → Evaluasi → Perbaikan Keputusan.",

"## Kesimpulan",

"EDA lanjutan membantu mengubah data mentah menjadi informasi yang lebih mudah dipahami. Kombinasi trend analysis, segmentasi, outlier analysis, dan dashboard dapat menjadi dasar untuk analisis yang lebih terstruktur."

]

create_pdf(
    "Business Recommendation.pdf",
    "Business Recommendation",
    recommendation_content
)

# ============================================================
# 15. SIMPAN DATA HASIL ANALISIS
# ============================================================

monthly.to_csv(
    "Trend_Analysis_Result.csv",
    index=False
)

segment_1.to_csv(
    "Segment_Species.csv",
    index=False
)

segment_2.to_csv(
    "Segment_Sepal_Length.csv",
    index=False
)

segment_3.to_csv(
    "Segment_Sepal_Width.csv",
    index=False
)

segment_4.to_csv(
    "Segment_Petal_Length.csv",
    index=False
)

segment_5.to_csv(
    "Segment_Petal_Width.csv",
    index=False
)

# ============================================================
# 16. HASIL AKHIR
# ============================================================

print("\n")
print("=" * 70)
print("🎉 WEEK 7 SELESAI")
print("=" * 70)

files = [
    "Dashboard.xlsx",
    "Trend Analysis.pdf",
    "Segment Analysis.pdf",
    "Outlier Analysis.pdf",
    "Insight Report.pdf",
    "Business Recommendation.pdf",
    "Trend_Analysis_Result.csv",
    "Segment_Analysis_Summary.csv",
    "Outlier_Analysis_Summary.csv",
    "Trend_Line_Chart.png",
    "Trend_Area_Chart.png",
    "Segment_Stacked_Bar.png",
    "Segment_Clustered_Bar.png",
    "Outlier_Boxplot.png",
    "Outlier_Distribution.png",
    "Pattern_Scatter_Plot.png",
    "Dashboard_Sales_Chart.png",
    "Dashboard_Category_Chart.png",
    "Dashboard_Region_Chart.png"
]

print("\nFILE YANG BERHASIL DIBUAT:")

for i, file in enumerate(files, 1):

    if os.path.exists(file):

        print(f"✅ {file}")

print("\n" + "=" * 70)
print("DELIVERABLE WEEK 7")
print("=" * 70)

print("""
✅ Dashboard.xlsx
✅ Trend Analysis.pdf
✅ Segment Analysis.pdf
✅ Outlier Analysis.pdf
✅ Insight Report.pdf
✅ Business Recommendation.pdf
⬜ Notebook.ipynb

Notebook.ipynb:
Gunakan notebook tempat kode ini dijalankan, kemudian pilih
File → Save and Checkpoint / Save Notebook.
""")

print("\nSemua analisis Week 7 sudah selesai.")
```

```python
# ============================================================
# WEEK 7 - FINAL PROJECT EDA
# FINAL PRESENTATION
# Dataset: Iris
# ============================================================

# Install library jika belum tersedia
import sys
!{sys.executable} -m pip install python-pptx -q

# ============================================================
# IMPORT
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris

from pptx import Presentation
from pptx.util import Inches, Pt
from pptx.enum.text import PP_ALIGN
from pptx.enum.shapes import MSO_SHAPE
from pptx.dml.color import RGBColor

print("=" * 70)
print("FINAL PROJECT EDA - WEEK 7")
print("=" * 70)

# ============================================================
# 1. LOAD DATASET
# ============================================================

iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=[
        "sepal_length",
        "sepal_width",
        "petal_length",
        "petal_width"
    ]
)

df["species"] = pd.Categorical.from_codes(
    iris.target,
    iris.target_names
)

numeric_cols = [
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width"
]

print("\nDataset berhasil dimuat.")
print("Jumlah baris :", len(df))
print("Jumlah kolom :", len(df.columns))

# ============================================================
# 2. STATISTIK
# ============================================================

stats = df[numeric_cols].describe().T

stats["median"] = df[numeric_cols].median()

correlation = df["petal_length"].corr(
    df["petal_width"]
)

species_count = (
    df["species"]
    .value_counts()
    .sort_index()
)

print("\nJumlah data setiap species:")
print(species_count)

print("\nKorelasi Petal Length dan Petal Width:")
print(round(correlation, 3))

# ============================================================
# 3. BUAT VISUALISASI
# ============================================================

# ------------------------------------------------------------
# VISUAL 1 - DISTRIBUSI PETAL LENGTH
# ------------------------------------------------------------

plt.figure(figsize=(9, 5))

for species in df["species"].unique():

    data = df[
        df["species"] == species
    ]["petal_length"]

    plt.hist(
        data,
        bins=10,
        alpha=0.5,
        label=str(species)
    )

plt.title("Distribusi Petal Length berdasarkan Species")
plt.xlabel("Petal Length")
plt.ylabel("Frekuensi")
plt.legend()

plt.tight_layout()

hist_path = "Final_Distribution.png"
plt.savefig(hist_path, dpi=150)
plt.close()

# ------------------------------------------------------------
# VISUAL 2 - SCATTER PETAL
# ------------------------------------------------------------

plt.figure(figsize=(9, 5))

for species in df["species"].unique():

    data = df[
        df["species"] == species
    ]

    plt.scatter(
        data["petal_length"],
        data["petal_width"],
        label=str(species)
    )

plt.title("Hubungan Petal Length dan Petal Width")
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()

scatter_path = "Final_Scatter.png"
plt.savefig(scatter_path, dpi=150)
plt.close()

# ------------------------------------------------------------
# VISUAL 3 - RATA-RATA SPECIES
# ------------------------------------------------------------

species_mean = (
    df.groupby("species", observed=True)[
        ["sepal_length",
         "sepal_width",
         "petal_length",
         "petal_width"]
    ]
    .mean()
)

plt.figure(figsize=(10, 5))

species_mean.plot(
    kind="bar"
)

plt.title("Rata-rata Variabel berdasarkan Species")
plt.xlabel("Species")
plt.ylabel("Rata-rata")
plt.xticks(rotation=0)

plt.tight_layout()

bar_path = "Final_Species_Mean.png"
plt.savefig(bar_path, dpi=150)
plt.close()

# ------------------------------------------------------------
# VISUAL 4 - CORRELATION HEATMAP
# ------------------------------------------------------------

corr_matrix = df[numeric_cols].corr()

plt.figure(figsize=(7, 6))

plt.imshow(
    corr_matrix,
    interpolation="nearest"
)

plt.colorbar()

plt.xticks(
    range(len(numeric_cols)),
    numeric_cols,
    rotation=30
)

plt.yticks(
    range(len(numeric_cols)),
    numeric_cols
)

plt.title("Correlation Matrix Dataset Iris")

for i in range(len(numeric_cols)):

    for j in range(len(numeric_cols)):

        plt.text(
            j,
            i,
            f"{corr_matrix.iloc[i, j]:.2f}",
            ha="center",
            va="center"
        )

plt.tight_layout()

heatmap_path = "Final_Correlation.png"
plt.savefig(heatmap_path, dpi=150)
plt.close()

print("\nVisualisasi berhasil dibuat.")

# ============================================================
# 4. DETEKSI OUTLIER
# ============================================================

outlier_summary = []

for col in numeric_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    mask = (
        (df[col] < lower) |
        (df[col] > upper)
    )

    outlier_summary.append({
        "Variable": col,
        "Outlier": int(mask.sum())
    })

outlier_df = pd.DataFrame(
    outlier_summary
)

print("\nOutlier:")
print(outlier_df)

# ============================================================
# 5. MEMBUAT POWERPOINT
# ============================================================

prs = Presentation()

prs.slide_width = Inches(13.333)
prs.slide_height = Inches(7.5)

# ============================================================
# WARNA
# ============================================================

NAVY = RGBColor(35, 48, 70)
PURPLE = RGBColor(104, 78, 145)
LIGHT = RGBColor(245, 243, 250)
WHITE = RGBColor(255, 255, 255)
DARK = RGBColor(45, 45, 55)
GRAY = RGBColor(110, 110, 120)

# ============================================================
# FUNGSI
# ============================================================

def add_background(slide):

    shape = slide.shapes.add_shape(
        MSO_SHAPE.RECTANGLE,
        0,
        0,
        prs.slide_width,
        prs.slide_height
    )

    shape.fill.solid()
    shape.fill.fore_color.rgb = LIGHT

    shape.line.fill.background()

    slide.shapes._spTree.remove(shape._element)
    slide.shapes._spTree.insert(2, shape._element)


def add_title(slide, title, subtitle=None):

    box = slide.shapes.add_textbox(
        Inches(0.7),
        Inches(0.35),
        Inches(12),
        Inches(0.7)
    )

    tf = box.text_frame

    p = tf.paragraphs[0]

    p.text = title

    p.font.size = Pt(27)
    p.font.bold = True
    p.font.color.rgb = NAVY

    if subtitle:

        box2 = slide.shapes.add_textbox(
            Inches(0.72),
            Inches(1.05),
            Inches(11.8),
            Inches(0.5)
        )

        p2 = box2.text_frame.paragraphs[0]

        p2.text = subtitle

        p2.font.size = Pt(12)
        p2.font.color.rgb = GRAY


def add_text(
    slide,
    text,
    x,
    y,
    w,
    h,
    size=18,
    bold=False,
    color=DARK,
    align=PP_ALIGN.LEFT
):

    box = slide.shapes.add_textbox(
        Inches(x),
        Inches(y),
        Inches(w),
        Inches(h)
    )

    tf = box.text_frame
    tf.word_wrap = True

    p = tf.paragraphs[0]

    p.text = text

    p.font.size = Pt(size)
    p.font.bold = bold
    p.font.color.rgb = color

    p.alignment = align

    return box


def add_bullet_list(
    slide,
    items,
    x,
    y,
    w,
    h,
    size=17
):

    box = slide.shapes.add_textbox(
        Inches(x),
        Inches(y),
        Inches(w),
        Inches(h)
    )

    tf = box.text_frame
    tf.word_wrap = True

    for i, item in enumerate(items):

        if i == 0:
            p = tf.paragraphs[0]
        else:
            p = tf.add_paragraph()

        p.text = item
        p.font.size = Pt(size)
        p.font.color.rgb = DARK
        p.space_after = Pt(10)
        p.level = 0

    return box


def add_image(
    slide,
    path,
    x,
    y,
    w=None,
    h=None
):

    if not os.path.exists(path):
        return

    if w and h:

        slide.shapes.add_picture(
            path,
            Inches(x),
            Inches(y),
            width=Inches(w),
            height=Inches(h)
        )

    elif w:

        slide.shapes.add_picture(
            path,
            Inches(x),
            Inches(y),
            width=Inches(w)
        )

    else:

        slide.shapes.add_picture(
            path,
            Inches(x),
            Inches(y)
        )


# ============================================================
# SLIDE 1 - COVER
# ============================================================

slide = prs.slides.add_slide(
    prs.slide_layouts[6]
)

add_background(slide)

add_text(
    slide,
    "FINAL PROJECT EDA",
    0.8,
    1.5,
    11.8,
    0.8,
    size=34,
    bold=True,
    color=NAVY,
    align=PP_ALIGN.CENTER
)

add_text(
    slide,
    "Exploratory Data Analysis Lanjutan",
    0.8,
    2.35,
    11.8,
    0.6,
    size=24,
    bold=True,
    color=PURPLE,
    align=PP_ALIGN.CENTER
)

add_text(
    slide,
    "Dataset Iris 🌸",
    0.8,
    3.1,
    11.8,
    0.7,
    size=22,
    color=DARK,
    align=PP_ALIGN.CENTER
)

add_text(
    slide,
    "Trend • Segmentasi • Outlier • Pattern • Insight • Recommendation",
    1.0,
    4.2,
    11.3,
    0.6,
    size=15,
    color=GRAY,
    align=PP_ALIGN.CENTER
)

add_text(
    slide,
    "Week 7 EDA Advanced",
    0.8,
    6.2,
    11.8,
    0.4,
    size=13,
    color=GRAY,
    align=PP_ALIGN.CENTER
)

# ============================================================
# SLIDE 2 - GAMBARAN DATASET
# ============================================================

slide = prs.slides.add_slide(
    prs.slide_layouts[6]
)

add_background(slide)

add_title(
    slide,
    "1. Gambaran Dataset",
    "Mengenal struktur dan karakteristik data"
)

add_bullet_list(
    slide,
    [
        "Dataset yang digunakan adalah Iris Dataset.",
        "Jumlah data: 150 observasi.",
        "Jumlah species: 3, yaitu setosa, versicolor, dan virginica.",
        "Setiap species memiliki 50 observasi.",
        "Terdapat 4 variabel numerik: sepal length, sepal width, petal length, dan petal width.",
        "Dataset tidak memiliki missing value."
    ],
    0.9,
    1.6,
    11.5,
    4.8,
    size=18
)

# ============================================================
# SLIDE 3 - STATISTIK DESKRIPTIF
# ============================================================

slide = prs.slides.add_slide(
    prs.slide_layouts[6]
)

add_background(slide)

add_title(
    slide,
    "2. Statistik Deskriptif",
    "Ringkasan karakteristik variabel numerik"
)

# Buat tabel
rows = 5
cols = 5

table = slide.shapes.add_table(
    rows,
    cols,
    Inches(1.0),
    Inches(1.7),
    Inches(11.3),
    Inches(3.8)
).table

headers = [
    "Variabel",
    "Mean",
    "Median",
    "Min",
    "Max"
]

for j, header in enumerate(headers):

    cell = table.cell(0, j)

    cell.text = header

    for p in cell.text_frame.paragraphs:
        p.font.bold = True
        p.font.size = Pt(14)
        p.font.color.rgb = WHITE

    cell.fill.solid()
    cell.fill.fore_color.rgb = PURPLE


for i, col in enumerate(numeric_cols, 1):

    values = [
        col,
        f"{df[col].mean():.2f}",
        f"{df[col].median():.2f}",
        f"{df[col].min():.2f}",
        f"{df[col].max():.2f}"
    ]

    for j, value in enumerate(values):

        cell = table.cell(i, j)

        cell.text = str(value)

        for p in cell.text_frame.paragraphs:
            p.font.size = Pt(13)
            p.font.color.rgb = DARK

add_text(
    slide,
    "Statistik menunjukkan bahwa setiap variabel memiliki rentang nilai yang berbeda.",
    1.0,
    5.8,
    11.3,
    0.7,
    size=16,
    color=GRAY,
    align=PP_ALIGN.CENTER
)

# ============================================================
# SLIDE 4 - VISUALISASI UTAMA
# ============================================================

slide = prs.slides.add_slide(
    prs.slide_layouts[6]
)

add_background(slide)

add_title(
    slide,
    "3. Visualisasi Utama",
    "Melihat distribusi dan pola data secara visual"
)

add_image(
    slide,
    hist_path,
    0.7,
    1.45,
    w=6.0
)

add_image(
    slide,
    scatter_path,
    6.75,
    1.45,
    w=6.0
)

add_text(
    slide,
    "Distribusi Petal Length",
    1.2,
    6.35,
    5.0,
    0.4,
    size=14,
    bold=True,
    align=PP_ALIGN.CENTER
)

add_text(
    slide,
    "Petal Length vs Petal Width",
    7.2,
    6.35,
    5.0,
    0.4,
    size=14,
    bold=True,
    align=PP_ALIGN.CENTER
)

# ============================================================
# SLIDE 5 - CORRELATION
# ============================================================

slide = prs.slides.add_slide(
    prs.slide_layouts[6]
)

add_background(slide)

add_title(
    slide,
    "3. Visualisasi Utama — Correlation",
    "Hubungan antarvariabel numerik"
)

add_image(
    slide,
    heatmap_path,
    0.8,
    1.3,
    w=6.0
)

add_text(
    slide,
    f"Korelasi terkuat:\n\nPetal Length ↔ Petal Width\n\nr = {correlation:.3f}",
    7.2,
    2.0,
    5.0,
    2.8,
    size=22,
    bold=True,
    color=PURPLE,
    align=PP_ALIGN.CENTER
)

add_text(
    slide,
    "Nilai korelasi yang tinggi menunjukkan hubungan linear yang sangat kuat. Namun, korelasi tidak berarti sebab-akibat.",
    7.0,
    4.8,
    5.4,
    1.5,
    size=15,
    color=DARK,
    align=PP_ALIGN.CENTER
)

# ============================================================
# SLIDE 6 - INSIGHT PENTING
# ============================================================

slide = prs.slides.add_slide(
    prs.slide_layouts[6]
)

add_background(slide)

add_title(
    slide,
    "4. Insight Penting",
    "Temuan utama dari proses EDA"
)

add_bullet_list(
    slide,
    [
        "Dataset memiliki 150 observasi dan 3 species.",
        "Distribusi data antar-species seimbang: masing-masing 50 data.",
        "Tidak ditemukan missing value.",
        "Petal length dan petal width memiliki korelasi paling kuat, sekitar 0,96.",
        "Ukuran petal memberikan pola pemisahan species yang lebih jelas.",
        "Setosa cenderung memiliki ukuran petal paling kecil dibanding dua species lainnya.",
        "Segmentasi dapat membantu mengelompokkan data berdasarkan karakteristik tertentu.",
        "Outlier perlu diperiksa sebelum dianggap sebagai kesalahan."
    ],
    0.8,
    1.45,
    11.8,
    5.5,
    size=16
)

# ============================================================
# SLIDE 7 - PERMASALAHAN
# ============================================================

slide = prs.slides.add_slide(
    prs.slide_layouts[6]
)

add_background(slide)

add_title(
    slide,
    "5. Permasalahan yang Ditemukan",
    "Keterbatasan dan hal yang perlu diperhatikan"
)

add_bullet_list(
    slide,
    [
        "Dataset Iris bukan dataset bisnis sehingga tidak memiliki data penjualan.",
        "Tidak terdapat kolom tanggal sehingga trend waktu tidak dapat dianalisis secara nyata.",
        "Tidak terdapat wilayah, produk, pelanggan, atau kategori bisnis.",
        "Beberapa analisis Week 7 membutuhkan data simulasi atau data turunan.",
        "Outlier tidak boleh langsung dihapus karena belum tentu merupakan kesalahan.",
        "Korelasi tinggi tidak dapat digunakan untuk menyimpulkan hubungan sebab-akibat."
    ],
    0.9,
    1.6,
    11.5,
    4.9,
    size=18
)

# ============================================================
# SLIDE 8 - PELUANG BISNIS
# ============================================================

slide = prs.slides.add_slide(
    prs.slide_layouts[6]
)

add_background(slide)

add_title(
    slide,
    "6. Peluang Bisnis",
    "Contoh penerapan hasil EDA pada data bisnis nyata"
)

add_bullet_list(
    slide,
    [
        "Segmentasi pelanggan berdasarkan karakteristik dan perilaku.",
        "Segmentasi produk berdasarkan performa penjualan.",
        "Pemantauan KPI melalui dashboard.",
        "Analisis trend untuk mengetahui pertumbuhan dan penurunan performa.",
        "Deteksi transaksi tidak biasa menggunakan analisis outlier.",
        "Penggunaan variabel yang paling informatif untuk mendukung pengambilan keputusan."
    ],
    0.9,
    1.55,
    11.5,
    5.0,
    size=18
)

# ============================================================
# SLIDE 9 - REKOMENDASI
# ============================================================

slide = prs.slides.add_slide(
    prs.slide_layouts[6]
)

add_background(slide)

add_title(
    slide,
    "7. Rekomendasi",
    "Langkah yang dapat diterapkan setelah EDA"
)

add_bullet_list(
    slide,
    [
        "Pastikan dataset bisnis memiliki atribut waktu, produk, pelanggan, wilayah, kategori, dan nilai transaksi.",
        "Gunakan segmentasi untuk memahami kelompok data yang berbeda.",
        "Validasi outlier sebelum melakukan penghapusan atau perubahan.",
        "Gunakan dashboard untuk memonitor KPI secara berkala.",
        "Fokuskan analisis pada variabel yang paling informatif.",
        "Lakukan evaluasi secara berkala agar keputusan tetap sesuai kondisi data terbaru."
    ],
    0.9,
    1.55,
    11.5,
    5.0,
    size=18
)

# ============================================================
# SLIDE 10 - KESIMPULAN
# ============================================================

slide = prs.slides.add_slide(
    prs.slide_layouts[6]
)

add_background(slide)

add_title(
    slide,
    "Kesimpulan",
    "Hasil akhir Exploratory Data Analysis"
)

add_text(
    slide,
    "EDA membantu memahami data sebelum digunakan untuk analisis atau pengambilan keputusan.",
    1.0,
    1.7,
    11.3,
    0.8,
    size=22,
    bold=True,
    color=NAVY,
    align=PP_ALIGN.CENTER
)

add_bullet_list(
    slide,
    [
        "Dataset berhasil dipahami melalui statistik deskriptif.",
        "Visualisasi membantu menemukan pola dan hubungan antarvariabel.",
        "Segmentasi membantu melihat karakteristik kelompok data.",
        "Outlier perlu divalidasi sebelum dilakukan tindakan.",
        "Dashboard dapat digunakan untuk monitoring pada data bisnis nyata.",
        "Hasil EDA dapat menjadi dasar untuk analisis dan keputusan berikutnya."
    ],
    1.2,
    2.8,
    10.8,
    3.2,
    size=18
)

# ============================================================
# SLIDE 11 - Q&A
# ============================================================

slide = prs.slides.add_slide(
    prs.slide_layouts[6]
)

add_background(slide)

add_text(
    slide,
    "TERIMA KASIH 🌸",
    0.8,
    2.0,
    11.8,
    0.8,
    size=34,
    bold=True,
    color=NAVY,
    align=PP_ALIGN.CENTER
)

add_text(
    slide,
    "Ada pertanyaan?",
    0.8,
    3.0,
    11.8,
    0.7,
    size=26,
    bold=True,
    color=PURPLE,
    align=PP_ALIGN.CENTER
)

add_text(
    slide,
    "Final Project — Exploratory Data Analysis Lanjutan",
    0.8,
    4.2,
    11.8,
    0.5,
    size=15,
    color=GRAY,
    align=PP_ALIGN.CENTER
)

# ============================================================
# SIMPAN PPT
# ============================================================

output_file = "Final Presentation.pptx"

prs.save(output_file)

print("\n" + "=" * 70)
print("🎉 FINAL PRESENTATION BERHASIL DIBUAT")
print("=" * 70)

if os.path.exists(output_file):

    size_mb = os.path.getsize(output_file) / (1024 * 1024)

    print("\n✅ File:", output_file)
    print(f"✅ Ukuran: {size_mb:.2f} MB")
    print("✅ Jumlah slide:", len(prs.slides))

print("""
Isi presentasi:

1. Gambaran Dataset
2. Statistik Deskriptif
3. Visualisasi Utama
4. Insight Penting
5. Permasalahan yang Ditemukan
6. Peluang Bisnis
7. Rekomendasi
8. Kesimpulan
9. Q&A

Target presentasi: 10–15 menit.
""")

print("=" * 70)
```
