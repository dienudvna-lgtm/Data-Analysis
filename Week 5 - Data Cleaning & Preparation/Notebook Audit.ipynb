# Senin - Initial Data Audit

Week 5 - Data Cleaning & Preparation

```python

import pandas as pd

df = pd.read_csv("Iris.csv")

print("Lima data pertama:")
print(df.head())

print("\nUkuran dataset:", df.shape)
print("\nInformasi dataset:")
df.info()

print("\nMissing value:")
print(df.isnull().sum())

print("\nJumlah duplicate:", df.duplicated().sum())
print("\nStatistik deskriptif:")
print(df.describe())

audit = pd.DataFrame({
    "Kolom": df.columns,
    "Tipe Data": [str(df[c].dtype) for c in df.columns],
    "Missing Value": [df[c].isnull().sum() for c in df.columns],
    "Jumlah Unik": [df[c].nunique() for c in df.columns]
})

ringkasan = pd.DataFrame({
    "Keterangan": ["Jumlah Baris", "Jumlah Kolom",
                   "Missing Value", "Duplicate"],
    "Hasil": [len(df), len(df.columns),
              int(df.isnull().sum().sum()),
              int(df.duplicated().sum())]
})

with pd.ExcelWriter("Initial Data Audit.xlsx") as writer:
    ringkasan.to_excel(writer, sheet_name="Ringkasan", index=False)
    audit.to_excel(writer, sheet_name="Audit Per Kolom", index=False)

print("Initial Data Audit.xlsx berhasil dibuat.")
```
