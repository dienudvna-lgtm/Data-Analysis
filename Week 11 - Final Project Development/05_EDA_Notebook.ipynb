```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings("ignore")

df = pd.read_csv("../02_Data_Preparation/03_Clean_Dataset.csv")

print("Ukuran dataset:", df.shape)
display(df.head())
```


## 1. Data Overview

Tahap ini digunakan untuk melihat ukuran dataset, nama kolom,
tipe data, dan informasi awal sebelum melakukan analisis.


```python

print("Jumlah baris:", df.shape[0])
print("Jumlah kolom:", df.shape[1])

print("\nNama kolom:")
print(df.columns.tolist())

print("\nInformasi tipe data:")
df.info()
```


## 2. Data Profiling

Pengecekan dilakukan terhadap missing value, duplicate,
dan jumlah nilai unik setiap kolom.


```python

profiling = pd.DataFrame({
    "Column": df.columns,
    "Data_Type": df.dtypes.astype(str).values,
    "Missing_Value": df.isnull().sum().values,
    "Unique_Value": df.nunique().values
})

display(profiling)

print("Jumlah duplicate:", df.duplicated().sum())
```


## 3. Statistik Deskriptif


```python

display(df.describe(include="all").T)
```


## 4. Visualisasi 1 - Gender


```python

if "gender" in df.columns:
    df["gender"].value_counts().plot(kind="bar", figsize=(7,5))
    plt.title("Jumlah Pelanggan Berdasarkan Gender")
    plt.xlabel("Gender")
    plt.ylabel("Jumlah Pelanggan")
    plt.tight_layout()
    plt.show()
```


## 5. Visualisasi 2 - Kelompok Usia


```python

if "age_group" not in df.columns and "age" in df.columns:
    df["age_group"] = pd.cut(
        df["age"],
        bins=[0,18,25,35,45,55,100],
        labels=["≤18","19-25","26-35","36-45","46-55","56+"]
    )

if "age_group" in df.columns:
    df["age_group"].value_counts().sort_index().plot(
        kind="bar",
        figsize=(8,5)
    )
    plt.title("Distribusi Pelanggan Berdasarkan Kelompok Usia")
    plt.xlabel("Kelompok Usia")
    plt.ylabel("Jumlah Pelanggan")
    plt.tight_layout()
    plt.show()
```


## 6. Visualisasi 3 - Kategori Produk


```python

if "category" in df.columns:
    df["category"].value_counts().plot(
        kind="bar",
        figsize=(8,5)
    )
    plt.title("Jumlah Pembelian Berdasarkan Kategori Produk")
    plt.xlabel("Kategori Produk")
    plt.ylabel("Jumlah Pembelian")
    plt.tight_layout()
    plt.show()
```


## 7. Visualisasi 4 - Nilai Pembelian per Kategori


```python

if "category" in df.columns and "purchase_amount_usd" in df.columns:
    data = (
        df.groupby("category")["purchase_amount_usd"]
        .sum()
        .sort_values(ascending=False)
    )

    data.plot(kind="bar", figsize=(8,5))
    plt.title("Total Nilai Pembelian Berdasarkan Kategori")
    plt.xlabel("Kategori Produk")
    plt.ylabel("Total Purchase Amount")
    plt.tight_layout()
    plt.show()
```


## 8. Visualisasi 5 - Nilai Pembelian Berdasarkan Gender


```python

if "gender" in df.columns and "purchase_amount_usd" in df.columns:
    data = (
        df.groupby("gender")["purchase_amount_usd"]
        .sum()
        .sort_values(ascending=False)
    )

    data.plot(kind="bar", figsize=(7,5))
    plt.title("Total Nilai Pembelian Berdasarkan Gender")
    plt.xlabel("Gender")
    plt.ylabel("Total Purchase Amount")
    plt.tight_layout()
    plt.show()
```


## 9. Visualisasi 6 - Subscription


```python

if "subscription_status" in df.columns:
    df["subscription_status"].value_counts().plot(
        kind="bar",
        figsize=(7,5)
    )
    plt.title("Status Subscription Pelanggan")
    plt.xlabel("Subscription Status")
    plt.ylabel("Jumlah Pelanggan")
    plt.tight_layout()
    plt.show()
```


## 10. Visualisasi 7 - Metode Pembayaran


```python

if "payment_method" in df.columns:
    df["payment_method"].value_counts().plot(
        kind="bar",
        figsize=(8,5)
    )
    plt.title("Metode Pembayaran yang Digunakan")
    plt.xlabel("Payment Method")
    plt.ylabel("Jumlah Transaksi")
    plt.tight_layout()
    plt.show()
```


## 11. Visualisasi 8 - Jenis Pengiriman


```python

if "shipping_type" in df.columns:
    df["shipping_type"].value_counts().plot(
        kind="bar",
        figsize=(8,5)
    )
    plt.title("Jenis Pengiriman yang Dipilih")
    plt.xlabel("Shipping Type")
    plt.ylabel("Jumlah")
    plt.tight_layout()
    plt.show()
```


## 12. Visualisasi 9 - Review Rating


```python

if "review_rating" in df.columns:
    df["review_rating"].plot(
        kind="hist",
        bins=10,
        figsize=(8,5)
    )
    plt.title("Distribusi Review Rating")
    plt.xlabel("Review Rating")
    plt.ylabel("Jumlah Pelanggan")
    plt.tight_layout()
    plt.show()
```


## 13. Visualisasi 10 - Previous Purchases


```python

if "previous_purchases" in df.columns:
    df["previous_purchases"].plot(
        kind="hist",
        bins=15,
        figsize=(8,5)
    )
    plt.title("Distribusi Previous Purchases")
    plt.xlabel("Previous Purchases")
    plt.ylabel("Jumlah Pelanggan")
    plt.tight_layout()
    plt.show()
```


## 14. 10 Key Insights

Insight dibuat berdasarkan hasil analisis dataset.


```python

insights = []

if "gender" in df.columns:
    x = df["gender"].value_counts()
    insights.append(
        f"1. Gender dengan jumlah pelanggan terbanyak adalah {x.idxmax()} "
        f"dengan {x.max()} pelanggan."
    )

if "age_group" in df.columns:
    x = df["age_group"].value_counts()
    insights.append(
        f"2. Kelompok usia dengan jumlah pelanggan terbanyak adalah {x.idxmax()}."
    )

if "category" in df.columns:
    x = df["category"].value_counts()
    insights.append(
        f"3. Kategori produk dengan jumlah pembelian terbanyak adalah {x.idxmax()}."
    )

if "category" in df.columns and "purchase_amount_usd" in df.columns:
    x = df.groupby("category")["purchase_amount_usd"].sum()
    insights.append(
        f"4. Kategori dengan total nilai pembelian terbesar adalah {x.idxmax()}."
    )

if "gender" in df.columns and "purchase_amount_usd" in df.columns:
    x = df.groupby("gender")["purchase_amount_usd"].sum()
    insights.append(
        f"5. Gender dengan total nilai pembelian terbesar adalah {x.idxmax()}."
    )

if "subscription_status" in df.columns:
    x = df["subscription_status"].value_counts()
    insights.append(
        f"6. Status subscription yang paling banyak adalah {x.idxmax()}."
    )

if "payment_method" in df.columns:
    x = df["payment_method"].value_counts()
    insights.append(
        f"7. Metode pembayaran yang paling banyak digunakan adalah {x.idxmax()}."
    )

if "shipping_type" in df.columns:
    x = df["shipping_type"].value_counts()
    insights.append(
        f"8. Jenis pengiriman yang paling banyak dipilih adalah {x.idxmax()}."
    )

if "review_rating" in df.columns:
    insights.append(
        f"9. Rata-rata review rating pelanggan adalah "
        f"{df['review_rating'].mean():.2f}."
    )

if "previous_purchases" in df.columns:
    insights.append(
        f"10. Rata-rata previous purchases pelanggan adalah "
        f"{df['previous_purchases'].mean():.2f}."
    )

for insight in insights:
    print(insight)
```


## 15. 3 Key Findings

Tiga temuan utama digunakan sebagai dasar untuk analisis bisnis
dan penyusunan rekomendasi.


```python

print("KEY FINDING 1")
print(insights[0])

print("\nKEY FINDING 2")
print(insights[2])

print("\nKEY FINDING 3")
print(insights[3])
```
